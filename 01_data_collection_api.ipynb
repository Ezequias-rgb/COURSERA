# Lab 1: SpaceX API Data Collection

Request SpaceX launch records through REST API v4. If the live service is unavailable, transparently load the course snapshot previously collected from SpaceX. The current endpoint returned HTTP 525 during project setup.

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents, Path.cwd() / 'spacex-capstone']
PROJECT_DIR = next(path for path in candidates if (path / 'src' / 'spacex_capstone.py').is_file())
sys.path.insert(0, str(PROJECT_DIR / 'src'))
from spacex_capstone import collect_launch_data

launches, collection_metadata = collect_launch_data(timeout=(3, 6))
print(collection_metadata)
print('shape:', launches.shape)
display(launches.head())
```

```python
output_path = PROJECT_DIR / 'data' / 'processed' / 'api_collected_launches.csv'
output_path.parent.mkdir(parents=True, exist_ok=True)
launches.to_csv(output_path, index=False)
print('saved:', output_path)
```

## Data provenance and limitation

The live endpoint is attempted on every run. The included CSV is the IBM course snapshot of SpaceX data; it is used only when the live request fails and the failure is recorded in `collection_metadata`. The analysis labs use the paired 90-launch processed dataset for consistent results.
