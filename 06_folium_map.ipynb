# Lab 6: Interactive Visual Analytics with Folium

Build three maps: launch-site markers, clustered launch outcomes, and a focused proximity view. Distances are geodesic between recorded launch-site coordinates, not road-route distances.

```python
from pathlib import Path
import sys
from IPython.display import IFrame, display

candidates = [Path.cwd(), *Path.cwd().parents, Path.cwd() / 'spacex-capstone']
PROJECT_DIR = next(path for path in candidates if (path / 'src' / 'spacex_capstone.py').is_file())
sys.path.insert(0, str(PROJECT_DIR / 'src'))
from spacex_capstone import create_folium_maps

map_results = create_folium_maps(selected_site='KSC LC 39A')
display(map_results['site_summary'].round(3))
display(map_results['distance_table'].round(2))
for map_path in map_results['maps']:
    print(map_path.name)
    display(IFrame(src=str(map_path), width='100%', height=480))
```
