# Lab 7: Interactive Dashboard with Plotly Dash

The runnable app is `dashboard/app.py`. It implements the launch-site dropdown, success pie chart, payload range slider, and outcome scatter plot, plus summaries for the five visual-analysis questions.

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents, Path.cwd() / 'spacex-capstone']
PROJECT_DIR = next(path for path in candidates if (path / 'dashboard' / 'app.py').is_file())
sys.path.insert(0, str(PROJECT_DIR / 'dashboard'))
import app as spacex_dashboard

client = spacex_dashboard.app.server.test_client()
assert client.get('/').status_code == 200
print('Dash page status:', client.get('/').status_code)
print('Dropdown sites:', spacex_dashboard.site_names)
print('Payload slider:', spacex_dashboard.payload_min, 'to', spacex_dashboard.payload_max, 'kg')
for question, result in spacex_dashboard.dashboard_insights().items():
    print(f'--- {question} ---')
    display(result.round(3))
```

## Run the application

From the project root, run `python dashboard/app.py` and open `http://127.0.0.1:8050`. Booster categories are derived from the supplied `Block` column because the source dataset labels the booster version only as `Falcon 9`.
